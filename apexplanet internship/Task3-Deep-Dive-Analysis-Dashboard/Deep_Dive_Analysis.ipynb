Formula:
COUNT(show_id)
Business Meaning:
Total content available on Netflix.

KPI 2: Movie Percentage
Formula:
Movies / Total Titles × 100
Business Meaning:
Measures movie dominance.

KPI 3: Top Country Contribution
Formula:
Country Titles / Total Titles × 100
Business Meaning:
Measures geographic dominance.

KPI 4: Mature Audience Content
Formula:
TV-MA / Total Titles × 100
Business Meaning:
Measures mature audience targeting.

KPI 5: Peak Production Year
Formula:
Year with highest number of titles
Business Meaning:
Measures growth trend.

# Deep-Dive Analysis 1: Content Growth Trend

Objective:
Analyze how Netflix content production changed over time.

```python
import matplotlib.pyplot as plt
year_counts = df['release_year'].value_counts().sort_index()
plt.figure(figsize=(12,5))
year_counts.plot(kind='line', marker='o')
plt.title('Netflix Content Growth Trend')
plt.xlabel('Release Year')
plt.ylabel('Number of Titles')
plt.show()
```

### Insight

Netflix content production increased rapidly between 2012 and 2018.
The highest number of titles was released in 2018.
After 2018, content growth declined gradually.

# Deep-Dive Analysis 2: Audience Analysis

Objective:
Analyze which audience groups Netflix targets.

```python
rating_counts = df['rating'].value_counts().head(10)
plt.figure(figsize=(10,5))
rating_counts.plot(kind='bar')
plt.title('Audience Rating Distribution')
plt.xlabel('Rating')
plt.ylabel('Number of Titles')
plt.show()
```

### Insight

Netflix primarily targets mature audiences.
TV-MA and TV-14 content dominate the platform.

# Deep-Dive Analysis 3: Geographic Analysis

Objective:
Analyze geographic distribution of Netflix content.

```python
top_countries = df['country'].value_counts().head(10)
plt.figure(figsize=(10,5))
top_countries.plot(kind='bar')
plt.title('Top Content Producing Countries')
plt.xlabel('Country')
plt.ylabel('Number of Titles')
plt.show()
```

### Insight

The United States dominates Netflix content production,
followed by India and the United Kingdom.
